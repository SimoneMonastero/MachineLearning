# Tuple
A tuple is a container that stores objects, also of different type, in a specific order.

```python
tpl = (1, 2)
print(tpl)
print(type(tpl))
```

```python
tpl = (1, 'Hello')
print(tpl)
print(type(tpl))
```

The tuple class provides some methods including:

```python
tpl = (1, 2, 3, 4, 6, 2, 3, 1, 4, 5)
print(tpl)

print(tpl.count(3))             # Return the number of times a specific value appears in the tuple

print(tpl.index(4))             # Search for the first occurrence of the specific value and return its position
```

Tuples are **immutable**. We cannot access a specific element and change it.

```python
tpl[0] = 2
```

Tuples are often used as **function parameters**. This prevents the code within the function from changing the values of the parameters during the function call

# Dictionary
Dictionaries contain couples of **(key: value)**.

```python
mydict = {'CT': 'not performed', 'MRI': 3, 'ECG': 20}
print(mydict)
print(type(mydict))
```

The key-set contains **unique** values.

```python
mydict = {'CT': 'not performed', 'CT': 3, 'ECG': 20}
print(mydict)
print(type(mydict))
```

We can access a dictionary by using the unique key value.

```python
print(1 in mydict)          # Checking the existence of a key
print(mydict.keys())        # Checking the dictionary keys

print(mydict['CT'])         # Accessing dictionary by key
```

Dictionaries are **mutable**.

```python
mydict['EEG']= [1, 2, 3]     # Adding an element to a dictionary - note that I am adding a new couple (key, value)
print(mydict)

mydict['EEG']= 3             # Changing an element of dictionary
print(mydict)
```

# Set
A set is a list of **unique** objects. Python will ignore duplicate items.

```python
st = {'CT', 'MRI'}
print(st)
print(type(st))
```

Sets are **mutable**, but **unsorted**.

```python
# Adding a element to a set
st.add('ECG')
print(st)
```

```python
# Converting a list to a set
lst = ['CT', 'CT', 'MRI', 'ECG']
st = set(lst)
print(st) # Casting list to set will delete duplicate items
```
