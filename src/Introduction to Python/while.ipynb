# while loop   
The **while** statement is used for repeated execution as long as an expression is true.

```python
a = False
b = 0
while a:
    b +=1
print(b)
```

```python
b = 0
while b<10:
    b = b+1
    print(b)
```

In the following, we will perform some exercises.

```python
### Exercise 1
# Starting from an empty list, add elements to it until its lenght is equal to 10
lst = []
c = 0;
while len(lst)<10:
  c = c+1;
  lst.append(c)
  print(lst)
```

```python
### Exercise 2
# Using a while loop, print only the first three items of the following tuple
myTuple = (10,20,30,40,50,60)
c = 0
while c < 3:
  print(myTuple[c])
  c = c+1;
```

```python
### Exercise 3
# Find the sum of numbers in a list using the while loop
myList = [23,45,12,10,25]
c = 0
sum = 0
while c < len(myList):
  sum = sum + myList[c]
  c += 1

print('La somma è: ', sum)
```
