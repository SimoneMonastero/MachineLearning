# IF conditional statement
Python presents two important type of control flow tools: **conditional statements** and **loops**.  
A **conditional statement** is a type of control structure that makes decision by analyzing the values of some variable. The main conditional statements in Python are: *if*, *if-else*, *if-elif-else*.   

The syntax of the *if* is:
```python
if condition:     # the header evaluates to a boolean object
    istructions   # the suite
```

The syntax of the *if-else* is:
```python
if condition:
    istructions
else:
    instructions
```

The syntax of the *if-elif-else* is:
```python
if condition:
    istructions
elif condition:
    instructions
else:
    instructions
```

In the previos syntax and in the following examples, note the **indentation**. Indentation is mandatory in Python to define the block of compund statements. Indentation is achieved by using tab or a constant number of white spaces.

````python
# if example
lesson = "Biomedical Signal Processing and Machine Learning"
if lesson == "Biomedical Signal Processing and Machine Learning":
    print("Remember to solve the exercises!")
````

````python
# if-else example
lesson = "Biomedical Signal Processing and Machine Learning"
if lesson == "Biomedical Signal Processing and Machine Learning":
    print("Remember to solve the exercises!")
else:
    print("Lunch time!")
````

````python
# if-elif-else example
lesson = "Biomedical Signal Processing and Machine Learning - Module 1"

if lesson == "Biomedical Signal Processing and Machine Learning - Module 2":
    print("Remember to solve signal processing exercises!")
elif lesson == 'Biomedical Signal Processing and Machine Learning - Module 1':
    print("Remember to solve machine learning exercises!")
else:
    print("Lunch time!")
````
