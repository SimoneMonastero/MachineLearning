# for loop
A Python **for loop** is used to iterate through an iterable (a process called *iterating*). **for loops** are used to repeat a block of code for a fixed number of times. The for-loop is always used in combination with an iterable object, like a list or a range. The Python **for statement** iterates over the members of a sequence in order (for more details click [here](https://wiki.python.org/moin/ForLoop)).

````python
str = "Python"     # the iterable object is a string
for char in str:
    print (char)
````

````python
lst = ["Python", "Matlab", "C", "C++", "C#"]     # the iterable object is a list
for language in lst:
    print (language)
````

````python
tpl = ("Python", "Matlab", "C", "C++", "C#")      # the iterable object is a tuple
for language in tpl:
    print (language)
````

````python
dict = {"Language": ["Python", "Matlab", "C", "C++", "C#"], "Readability": ["high", "high", "low", "low", "low"]}
for key in dict.keys():                           # the iterable object is the dictionary's keys'set
    print (key, dict[key])
````

````python
st = {"Python", "Matlab", "C", "C++", "C#"}      # the iterable object is a set
for language in st:
    print (language)
````

````python
count = 0
for i in range(0,10):
    count = count + i # count += i
    print(count)
````

The **range type** represents an immutable sequence of numbers and is commonly used for looping a specific number of times in for loops (for more details click [here](https://docs.python.org/3/library/stdtypes.html#typesseq)).  
The *range* syntax is:
```python
range([start, ]stop[, step])
```
*stop* element is always excluded!

````python
# Range syntax example 1
for i in range(10):
    print(i)
````

````python
# Range syntax example 2
for i in range(4,10):
    print(i)
````

````python
# Range syntax example 3
for i in range(-6,10,2):
    print(i)
````

````python
### Exercise 8
# Iterate the interval [1,50]. For multiples of three print "Fizz" instead of the number and for multiples of five print "Buzz". For numbers that are multiples of three and five, print "FizzBuzz".
for i in range(1,51):
  if i%3==0 and i%5==0:
    print("FizzBuzz")
  elif i%3==0:
    print("Fizz")
  elif i%5==0:
    print("Buzz")
````

````python
### Exercise 9
# Check the validity of a password string, which must contain:
# At least 1 letter between [a-z] and 1 letter between [A-Z]
# At least 1 number between [0-9]
# At least 1 character from [$#@]
# Minimum length 6 characters
# Maximum length 16 characters
password = input("Enter a password: ")
if len(password) < 6 or len(password) > 16:
  print("Invalid password")
  exit()
if not any(char.isdigit() for char in password):
  print("Invalid password")
  exit()
if not any(char.isupper() for char in password):
  print("Invalid password")
  exit()
if not any(char.islower() for char in password):
  print("Invalid password")
  exit()
if not any(char in ["$", "#", "@"] for char in password):
  print("Invalid password")
  exit()
  print("Valid password")
````
