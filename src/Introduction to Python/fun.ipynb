# Functions   
A function is a block of code which runs only when it is called. In order to use a function, we should **define** it and then **call** it.   
```python
# Function definition
def func_name(par1, par2, par3, ...):
    instruction
    return out1, out2, oput3, ...

# Function call
out1,out2, ou3, ... = func_name(par1, par2, par3, ...)
```

````python
def square(x):
    y = x**2
    return y

var = 2
out = square(var)
print(out)
````

In the **definition** phase, you can add as many parameters as you want, just separate them with a comma. We can also define default values for parameters, by adding a specific value (and so type) in the function definition.   
In the **call** phase, the parameters for which the default is defined are optional, while the others are mandatory and must be passed by position or name.

````python
#Function definition
def pow(base, exp = 1):
    out = base**exp
    return out
````

````python
# Function call - mandatory parameter passed by position, no optional parameter passed --> the function uses the default value
var = 2
y = pow(var,2)
print(y)
````

````python
# Function call - mandatory and optional parameter passed by position
y = pow(var, 3)
print(y)
````

````python
# Function call - mandatory passed by name, no optional parameter passed --> the function uses the default value
y = pow(base = var)
print(y)
````

````python
# Function call - mandatory and optional parameter passed by name
y = pow(base = var, exp = 3)
print(y)
````

````python
# Function call - mandatory and optional parameter passed by name
y = pow(exp = 3, base = var)
print(y)
````

````python
# Function call - unknown optional parameter passed by name (TypeError)
y = pow(exp = 3, x = var)
print(y)
````
