# List

A **list** is a container that stores objects, also of different type, in a specific order.

```python
lst = [0, 1, 2, "hi"]
print(lst)
print(type(lst))
```

The list class provides numerous methods (for a comprehensive list go to the [offical page](https://docs.python.org/3/tutorial/datastructures.html)) including:

```python
lst = ["Welcome", 2023]
print(lst)

lst.append('!')             # Add an item to the end of the list.
print(lst)

lst.insert(1, ',')          # Insert an item at a given position.
print(lst)

lst.remove("Welcome")       # Remove the first item from the list whose value is equal to the parameter. It raises a ValueError if there is no such item.
print(lst)

lst = ['c', 'd', 'a']
print(lst)
lst.sort()                  # Sort the items of the list in place
print(lst)
```

Lists are **mutable**. We can access a specific element and change it.

```python
lst = ['c', 'd', 'a']
print(lst)
lst[2] = 'e'
print(lst)
```
